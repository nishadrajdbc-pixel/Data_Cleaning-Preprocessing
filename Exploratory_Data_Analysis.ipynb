```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df.head()
```

```python
df.describe()
df.median(numeric_only=True)
df.std(numeric_only=True)
```

```python
sns.histplot(df["Age"], kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.boxplot(x=df["Fare"])
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

```python
sns.pairplot(df[["Age","Fare","Pclass","Survived"]], hue="Survived")
plt.show()
```

```python
sns.countplot(x="Sex", hue="Survived", data=df)
plt.show()

#Female passengers had significantly higher survival rates compared to males, indicating priority during evacuation.
```

```python
sns.countplot(x="Pclass", hue="Survived", data=df)
plt.show()
# Passengers from 1st class had higher survival probability, suggesting socioeconomic advantage.
```
